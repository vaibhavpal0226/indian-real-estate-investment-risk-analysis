+ **Market Segmentation:** Map the supply of "Family-Sized" vs. "Bachelor/Studio" units to find market gaps.
+ **Scarcity Analysis:** Isolate "Rare Assets" (low-density homes) in high-density urban markets for long-term land appreciation.

##### __*Duplicates Handling*__
- When I checked for duplicates in primary key column i.e. 'ListingID', it is found that single 'ListingID' is use for two different property i.e having the same ID for a flat in Mumbai and a house in Delhi. ListingID should ideally point to one specific home as it is unique identifier.  
- The duplicate ListingIDs indicates that the data collection system reused IDs for different properties in different regions. So for finding actual duplicates I used "Composite Key Duplication Strategy" i.e. a row is only a duplicate if multiple columns match exactly (e.g., City + Locality + Price_INR + SuperBuiltUpArea_sqft).
- This strategy helps me to conclude that there are no duplicates present in the record even though "ListingID" is same.

```python
# Checking total nulls & percentage per column.
null_counts = df_filtered.isnull().sum()
null_pct = round((null_counts/len(df_filtered))*100,2)

missing_data = pd.DataFrame({'Total Nulls': null_counts, 'Percentage %': null_pct})
display(missing_data)
```

```python
# Dropping rows where both city and locality have NaN values
df_filtered_drop_null = df_filtered.dropna(subset=['city', 'locality'])

# Removing duplicates in locality only for mapping purpose
mapping = df_filtered_drop_null.drop_duplicates('locality')
# Making mapping dictionary
mapping_dict = pd.Series(mapping.city.values, index=mapping.locality.values).to_dict()
# Filling missing cities using the mapping
df_filtered['city'] = df_filtered['city'].fillna(df_filtered['locality'].map(mapping_dict))

# Checking how many values are filled
remaining_nulls = df_filtered['city'].isnull().sum()
print(f"Cities recovered: {50 - remaining_nulls}")
```

```python
# Imputing missing values for locality column.
df_filtered['locality'] = df_filtered['locality'].fillna('Unknown')
```

```python
# Imputing Missing Values for bathrooms column.
df_filtered['bathrooms'] = df_filtered.groupby('bhk')['bathrooms'].transform(lambda x: x.fillna(x.median()))
```

```python
# Imputing missing values for latitude and longitude column using city column.

# Before imputing with we are checking if outliers are present in latitiue and longitude column.
outliers = df_filtered[(df_filtered['latitude'] < 6) | (df_filtered['latitude'] > 37) | 
              (df_filtered['longitude'] < 68) | (df_filtered['longitude'] > 97)]
print(f"Properties with physically impossible Indian coordinates: {len(outliers)}")

# Imputing with mean
city_coordinates = df_filtered.groupby('city')[['latitude', 'longitude']].transform('mean')
df_filtered['latitude'] = df_filtered['latitude'].fillna(city_coordinates['latitude'])
df_filtered['longitude'] = df_filtered['longitude'].fillna(city_coordinates['longitude'])
```

```python
# Imputing missing values for reraid column using IsRERARegisterd column.
df_filtered['reraid'] = df_filtered['reraid'].str.strip().replace(['nan', 'None', ''], np.nan)
# Case A: RERA not registered.
df_filtered.loc[df_filtered['isreraregistered'] == 0, 'reraid'] = 'Not Applicable'
# Case B: RERA registered but ID not provided.
df_filtered.loc[(df_filtered['isreraregistered'] == 1) & (df_filtered['reraid'].isna()), 'reraid'] = 'ID Not Provided'
```

```python
# Imputing missing values for yearbuilt using ageyears column.
df_filtered['yearbuilt'] = df_filtered['yearbuilt'].replace(0, np.nan)
df_filtered['year_calculated'] = 2025 - df_filtered['ageyears']
df_filtered['yearbuilt'] = df_filtered['yearbuilt'].fillna(df_filtered['year_calculated'])

df_filtered.drop(columns=['year_calculated'], inplace=True)
```

```python
# Imputing missing values in parking column using mode (most frequent)

# First dropping off any NaN in parking column before finding most common.
parking_drop_null = df_filtered.dropna(subset=['parking']).groupby('buildingtype')['parking']

# Applying mode and making mapping dictionary.
parking_mapping = parking_drop_null.apply(lambda x: x.mode()[0]).to_dict() # .mode() returns list, [0] choose first most common

print("Parking Logic:")
for b_type, p_type in parking_mapping.items():
    print(f" - {b_type} usually has {p_type} parking")

df_filtered['parking'] = df_filtered['parking'].fillna(df_filtered['buildingtype'].map(parking_mapping))
```

```python
# Final check for missing values
df_filtered.isnull().sum()
```

##### __*Missing Data Imputation*__
- If the price is missing, we cannot fulfill our goal of "best price" or "affordability." If the city/locality is missing, we can't map it. It is found that 50 cities and 70 localities are missing.
- We cannot impute locality or price value if it is missing but we can impute the City using the Locality, provided that the same locality exists elsewhere in the dataset with a city name attached.
- Used cross-reference imputation to recover missing City values based on known Locality mappings, maximizing the dataset's geographic utility.
- For remaining 70 missing Localities, used 'Unknown' placeholder to preserve price and area data for aggregate city-level analysis.
- A 1-BHK usually has 1 bathroom, while a 4-BHK has 3 or 4. We have imputed the missing values using median based on the BHK count.
- We have calculated the average (mean) Latitude and Longitude for each city and use those values to fill the gaps. This ensures that even if we don't know the exact street, the property at least appears in the correct city on the map. If we impute it using locality column then it gave us 90% accuracy but since, for remaining 70 missing Localities, used 'Unknown' placeholder so we used "City-Centroid" Imputation Strategy.
- For filling missing values in 'reraid' column we used two cases: 
    + Case 1: if a property is not registered, it shouldn't have an ID. 
    + Case 2: If it is registered but the ID is missing, it should flag it as "ID Not Provided" rather than leaving it as a NaN.
- Some 'year of built' for properties were missing but the age of property was given so we imputed missing values for 'YearBuilt' using logic: "2025 - age of property"
- We impute missing values in parking column using "Mode" (the most frequent value) of parking for each building type. For example, a "High Rise" is much more likely to have "Basement" or "Covered" parking than a "Standalone Building.

### __Feature Engineering__
We will create calculated columns to supplement the raw data, allowing for a more granular and insightful analysis.

```python
# 1. Price per Sqft
df_filtered['price_per_sqft'] = df_filtered['price_inr'] / df_filtered['superbuiltuparea_sqft']
```

```python
# 2. Locality Avg Price per Sqft

# Calculate how many properties are in each locality
df_filtered['group_count'] = df_filtered.groupby(['city', 'locality', 'propertytype', 'buildingtype'])['price_per_sqft'].transform('count')

# Calculate both benchmarks
df_filtered['initial_benchmark'] = df_filtered.groupby(['city', 'locality', 'propertytype', 'buildingtype'])['price_per_sqft'].transform('median')
df_filtered['final_benchmark'] = df_filtered.groupby(['city', 'locality'])['price_per_sqft'].transform('median')

# Fallback: Use initial_benchmark if count > 3, otherwise use final_benchmark
df_filtered['locality_avg_price_per_sqft'] = np.where(df_filtered['group_count'] > 3, 
                                 df_filtered['initial_benchmark'], 
                                 df_filtered['final_benchmark'])
```

```python
# 3. Value Gap (how much specific property diverge from average market price)
df_filtered['price_deviation_pct'] = ((df_filtered['price_per_sqft'] - df_filtered['locality_avg_price_per_sqft']) / df_filtered['locality_avg_price_per_sqft'] ) * 100

# Dropping helper column
df_filtered.drop(columns=['initial_benchmark', 'final_benchmark', 'group_count'], inplace=True)
```

```python
# 4.Price category based on 'price_deviation_pct'

# Minimum and maximum of price deviation percentage
print(df_filtered['price_deviation_pct'].min(), df_filtered['price_deviation_pct'].max(), sep=',' ,end='\n\n')

bins = [-40, -15, 15, 35, 60]
labels = ['Underpriced', 'Fair Market Value', 'Premium', 'Overpriced']
df_filtered['price_category'] = pd.cut(df_filtered['price_deviation_pct'], bins=bins, labels=labels)
print(df_filtered['price_category'].value_counts())
```

```python
# 5. Land Overlap Risk
# First ensuring that locality in not unknown, since we have imputed missing values as 'Unknown'
real_coords_mask = (df_filtered['locality'] != 'Unknown')
coord_counts = df_filtered[real_coords_mask].groupby(['latitude', 'longitude'])['listingid'].transform('count')

df_filtered['land_overlap_risk'] = 'Low Risk'
df_filtered.loc[real_coords_mask & (coord_counts > 1), 'land_overlap_risk'] = 'High Risk'
df_filtered.loc[df_filtered['locality'] == 'Unknown', 'land_overlap_risk'] = 'Inexact Location'

print(df_filtered['land_overlap_risk'].value_counts())
```

```python
# 6. Structural Life Stage
def classify_life_stage(row):
    age = row['ageyears']
    building_type = str(row['buildingtype']).lower()
    
    # Apartments / High-Rises
    if 'rise' in building_type or 'community' in building_type:
        if age <= 10: return 'New'
        if age <= 25: return 'Mid-Life'
        if age <= 40: return 'Aging'
        return 'Redevelopment Needed'
    
    # Standalone Houses / Bungalow
    else:
        if age <= 15: return 'New'
        if age <= 35: return 'Mid-Life'
        if age <= 50: return 'Aging'
        return 'Redevelopment Needed'

df_filtered['structural_life_stage'] = df_filtered.apply(classify_life_stage, axis=1)
```

```python
# 7. Safety Ratings
def safety_checks(row):
    penalty_points = 0

    # RERA unregistered
    if row['isreraregistered'] == 'No': penalty_points += 50
    elif row['reraid'] == 'ID Not Provided': penalty_points += 20

    # Suspiciously low price (maybe scam)
    if row['price_deviation_pct'] <= -40: penalty_points += 25

    # Land integrity
    if row['land_overlap_risk'] == 'High Risk': penalty_points += 40
    elif row['land_overlap_risk'] == 'Inexact Location': penalty_points += 10

    # Aging checks
    if row['structural_life_stage'] == 'Redevelopment Needed': penalty_points += 40 # High risk of major repairs
    elif row['structural_life_stage'] == 'Aging': penalty_points += 20 # Medium risk, requires audit

    # Final Categorization
    if penalty_points >= 50: return 'High Risk'
    elif penalty_points >= 25: return 'Caution'
    return 'Verified & Safe'  

df_filtered['safety_ratings'] = df_filtered.apply(safety_checks, axis=1)
```

```python
# 8. Space Efficiency Grade (how effectively a building's total floor area is utilized.)
df_filtered['space_efficiency_ratio'] = (df_filtered['carpetarea_sqft'] / df_filtered['superbuiltuparea_sqft']) * 100

bins = [0,60,80,100]
df_filtered['space_efficieny_grade'] = pd.cut(df_filtered['space_efficiency_ratio'], bins=bins, labels=['Poor Efficiency', 'Standard', 'High Efficiency'])
df_filtered.drop(columns=['space_efficiency_ratio'], inplace=True)
```

```python
# 9. Family Segment
def family_wise(row):
    # A 'Family Ready' home has 2+ bhk, 2+ bathrooms, and reserved parking
    if row['bhk'] >= 3 and row['bathrooms'] >= 3 and str(row['parking']).lower() != 'none':
        return 'Large Family Choice'
    if row['bhk'] >= 2 and row['bathrooms'] >= 1:
        return 'Small Family/Couple'
    return 'Bachelor/Studio'

df_filtered['family_segment'] = df_filtered.apply(family_wise, axis=1)
```

```python
# 10. Modernity index
def classify_modernity(row):
    score = 0
    # Amenity points
    if row['amenitiescount'] > 12: score += 2
    elif row['amenitiescount'] > 7: score += 1
    
    # Parking points
    if str(row['parking']).lower() in ['covered', 'basement', 'stilt']:
        score += 1

    # High Floor factor (Modern high-rises)
    if row['totalfloors'] > 15: score += 1    

    if score >= 3: return 'Ultra-Modern/Smart'
    if score >= 1: return 'Standard Modern'
    return 'Basic/Traditional'

df_filtered['modernity_index'] = df_filtered.apply(classify_modernity, axis=1)
```

```python
# 11. Utility Balance (competition vs. confort)
df_filtered['bath_to_bhk_ratio'] = df_filtered['bathrooms'] / df_filtered['bhk']
df_filtered['parking_count'] = df_filtered['parking'].apply(lambda x:0 if str(x).lower() == 'none' else 1)

def investment_utility_signal(row):
    # Under-serviced (Bad for Investment)
    if row['bhk'] >= 3 and row['bath_to_bhk_ratio'] < 0.6: # A 2 BHK with 1 bathroom (ratio 0.5); high competition, reducing comfort.
        return 'Under-Serviced'
    
    # Luxury Balance (High Investment Potential)
    if row['bath_to_bhk_ratio'] >= 1.0 and row['parking_count'] >= 1:
        return 'High-Utility'
    
    return 'Standard Utility'

df_filtered['utility_balance'] = df_filtered.apply(investment_utility_signal, axis=1)
df_filtered.drop(columns=['bath_to_bhk_ratio', 'parking_count'], inplace=True)
```

```python
# 12. FSI Category
def estimate_fsi_impact(row):
    # High-rise buildings have high FSI (more people sharing the same land)
    if row['totalfloors'] > 20:
        return 'High FSI (Dense Urban)'
    elif row['totalfloors'] > 5:
        return 'Medium FSI (Suburban Multi-story)'
    else:
        return 'Low FSI (Spacious/Independent)'

df_filtered['fsi_category'] = df_filtered.apply(estimate_fsi_impact, axis=1)
```

```python
# 13. Low Density Premium
df_filtered['low_density_premium'] = np.where(
    (df_filtered['fsi_category'] == 'Low FSI (Spacious/Independent)') & 
    (df_filtered['price_category'] == 'Fair Market Value'), 
    'Rare Asset', 'Standard'
)
```

*By adding 10+ engineered features, we transformed raw numbers into "Investment Signals" that a buyer can actually understand.*
+ **price_per_sqft:** Standardized price unit; allows property price comparison.
+ **locality_avg_price_sqft:** The median price of similar properties in the same neighborhood; acts as the "Fair Market Value."
+ **price_deviation_pct:** The percentage difference between the listing price and the locality benchmark; identifies bargains or overpriced units.
+ **price_category:** Categorizes properties into 4 buckets (Underpriced to Overpriced) based on price quantiles.
+ **land_overlap_risk:** Geospatial check that flags multiple listings on the exact same coordinate to prevent fraud.
+ **structural_life_stage:** Classifies the building into stages like New, Mid-Life, Aging, and Redevelopment Needed based on age and building type.
+ **safety_rating:** A proprietary risk score (Verified, Caution, High Risk) combining RERA status, age, price suspicion, land overlap risk and lifespan of buildings.
+ **space_efficiency_grade:** Measures how much of the total area is actually usable (Carpet vs. Super Built-up ratio).
+ **family_segment:** Segments properties (Bachelor, Small Family, Large Family) based on BHK and bathroom counts.
+ **modernity_index:** Ranks properties by "Smart/Modern" features using amenity counts, parking types, and floor height.
+ **utility_balance:** A score determining if the ratio of bathrooms and parking to BHK is sufficient for high rental demand.
+ **fsi_category:** Categorized Floor Space Index into three segments (Low, Medium and High FSI) based on the total floors. More number of floors means dense population, because many people sharing same land.
+ **low_density_premium:** Flags "Rare Assets"—luxury properties with low building density in high-demand cities.
